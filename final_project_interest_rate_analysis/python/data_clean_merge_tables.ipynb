```python
import pandas as pd
```

```python
#getting monthly average for daily files

df = pd.read_csv("fed_funds_effective_rate.csv", parse_dates=['date'])
monthly_avg = (
    df
    .groupby(pd.Grouper(key="date", freq="M"))["fed_funds_effective_rate"]
    .mean()
    .reset_index()
)
monthly_avg.to_csv("fed_funds_effective_rate_avg.csv", index=False)
```

```python
df = pd.read_csv("fed_nasdaq100_index.csv", parse_dates=['date'])
df["nasdaq100"] = pd.to_numeric(df["nasdaq100"], errors='coerce')
monthly_avg = (
    df
    .groupby(pd.Grouper(key="date", freq="M"))["nasdaq100"]
    .mean()
    .reset_index()
)
monthly_avg.to_csv("fed_nasdaq100_index_avg.csv", index=False)
```

```python
#combining all csv's to 1

df1 = pd.read_csv('fed_tbill_3mo_yield.csv')
df2 = pd.read_csv('fed_libor_rate.csv')
df3 = pd.read_csv('fed_sofr_rate_avg.csv')
df4 = pd.read_csv('fed_funds_effective_rate_avg.csv')
df5 = pd.read_csv('fed_tsec_10yr_avg.csv')
df6 = pd.read_csv('moody_aaa_yields.csv')


df_merged = df1.merge(df2, on='date', how='left')\
               .merge(df3, on='date', how='left')\
               .merge(df4, on='date', how='left')\
               .merge(df5, on='date', how='left')\
                .merge(df6, on='date', how='left')

df_merged.to_csv('interest_rate_data_combined.csv', index=False)
```
